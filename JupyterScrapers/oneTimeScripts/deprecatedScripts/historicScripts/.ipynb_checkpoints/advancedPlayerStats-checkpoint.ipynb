```python
from sqlalchemy import create_engine, text
from requests.exceptions import ReadTimeout
from datetime import datetime, timedelta
import pandas as pd
import random
import time 
import json

from dotenv import load_dotenv
import os

load_dotenv()

DATABASE_URL = os.getenv("DATABASE_URL")
engine = create_engine(DATABASE_URL)

def calculate_adv_stats(game_id='0022301147', player_id=1630552):
    """
    Calculate advanced stats for a single player in a single game.
    
    Returns a dictionary with all calculated advanced statistics.
    """
    
    with engine.connect() as conn:
        query = text("""
            SELECT 
                p.*,
                -- Player's team stats (all of them)
                pt.team_pts, 
                pt.team_fgm,
                pt.team_fga, 
                pt.team_fg_pct,
                pt.team_fg3m,
                pt.team_fg3a,
                pt.team_fg3_pct,
                pt.team_ftm,
                pt.team_fta, 
                pt.team_ft_pct,
                pt.team_oreb, 
                pt.team_dreb, 
                pt.team_reb, 
                pt.team_ast, 
                pt.team_stl, 
                pt.team_blk, 
                pt.team_tov,
                pt.team_pf,
                pt.team_min,
                -- Opponent's team stats (all of them)
                ot.team_pts as opp_pts, 
                ot.team_fgm as opp_fgm,
                ot.team_fga as opp_fga,
                ot.team_fg_pct as opp_fg_pct,
                ot.team_fg3m as opp_fg3m,
                ot.team_fg3a as opp_fg3a,
                ot.team_fg3_pct as opp_fg3_pct,
                ot.team_ftm as opp_ftm,
                ot.team_fta as opp_fta,
                ot.team_ft_pct as opp_ft_pct,
                ot.team_oreb as opp_oreb,
                ot.team_dreb as opp_dreb,
                ot.team_reb as opp_reb,
                ot.team_ast as opp_ast,
                ot.team_stl as opp_stl,
                ot.team_blk as opp_blk,
                ot.team_tov as opp_tov,
                ot.team_pf as opp_pf,
                ot.team_min as opp_min
            FROM player_game_stats p
            INNER JOIN team_game_stats pt 
                ON p.game_id = pt.game_id AND p.team_id = pt.team_id
            INNER JOIN team_game_stats ot 
                ON p.game_id = ot.game_id AND p.team_id != ot.team_id
            WHERE p.game_id = :game_id AND p.player_id = :player_id;
        """)
        
        result = conn.execute(query, {"game_id": game_id, "player_id": player_id})
        row = result.fetchone()
        
        if not row:
            print(f"No data found for game_id={game_id}, player_id={player_id}")
            return None
        
        # Convert row to dictionary for easier access
        data = dict(row._mapping)
    
    # Extract player stats
    player_min = data['min'] or 0
    player_pts = data['pts'] or 0
    player_fgm = data['fgm'] or 0
    player_fga = data['fga'] or 0
    player_fg3m = data['fg3m'] or 0
    player_ftm = data['ftm'] or 0
    player_fta = data['fta'] or 0
    player_oreb = data['oreb'] or 0
    player_dreb = data['dreb'] or 0
    player_reb = data['reb'] or 0
    player_ast = data['ast'] or 0
    player_stl = data['stl'] or 0
    player_blk = data['blk'] or 0
    player_tov = data['tov'] or 0
    player_pf = data['pf'] or 0
    
    # Extract team stats
    team_min = data['team_min'] or 0
    team_pts = data['team_pts'] or 0
    team_fgm = data['team_fgm'] or 0
    team_fga = data['team_fga'] or 0
    team_fta = data['team_fta'] or 0
    team_oreb = data['team_oreb'] or 0
    team_dreb = data['team_dreb'] or 0
    team_reb = data['team_reb'] or 0
    team_ast = data['team_ast'] or 0
    team_tov = data['team_tov'] or 0
    
    # Extract opponent stats
    opp_min = data['opp_min'] or 0
    opp_pts = data['opp_pts'] or 0
    opp_fgm = data['opp_fgm'] or 0
    opp_fga = data['opp_fga'] or 0
    opp_fta = data['opp_fta'] or 0
    opp_oreb = data['opp_oreb'] or 0
    opp_dreb = data['opp_dreb'] or 0
    opp_reb = data['opp_reb'] or 0
    opp_tov = data['opp_tov'] or 0
    
    # Calculate advanced stats
    adv_stats = {}
    
    # ==================== SIMPLE STATS (Player Only) ====================
    
    # Effective Field Goal Percentage
    adv_stats['efg_pct'] = (player_fgm + 0.5 * player_fg3m) / player_fga if player_fga > 0 else None
    
    # True Shooting Percentage
    adv_stats['ts_pct'] = player_pts / (2 * (player_fga + 0.44 * player_fta)) if (player_fga + player_fta) > 0 else None
    
    # Assist to Turnover Ratio
    adv_stats['ast_tov'] = player_ast / player_tov if player_tov > 0 else None
    
    # ==================== POSSESSIONS ====================
    
    # Team Possessions
    team_poss = 0.96 * (team_fga + team_tov + 0.44 * team_fta - team_oreb)
    adv_stats['team_poss'] = team_poss
    
    # Opponent Possessions
    opp_poss = 0.96 * (opp_fga + opp_tov + 0.44 * opp_fta - opp_oreb)
    adv_stats['opp_poss'] = opp_poss
    
    # ==================== USAGE & PERCENTAGE STATS ====================
    
    # Usage Percentage
    if player_min > 0 and team_min > 0:
        adv_stats['usg_pct'] = 100 * (
            (player_fga + 0.44 * player_fta + player_tov) * (team_min / 5)
        ) / (
            player_min * (team_fga + 0.44 * team_fta + team_tov)
        )
    else:
        adv_stats['usg_pct'] = None
    
    # Offensive Rebound Percentage
    if player_min > 0 and team_min > 0 and (team_oreb + opp_dreb) > 0:
        adv_stats['oreb_pct'] = 100 * (player_oreb * (team_min / 5)) / (
            player_min * (team_oreb + opp_dreb)
        )
    else:
        adv_stats['oreb_pct'] = None
    
    # Defensive Rebound Percentage
    if player_min > 0 and team_min > 0 and (team_dreb + opp_oreb) > 0:
        adv_stats['dreb_pct'] = 100 * (player_dreb * (team_min / 5)) / (
            player_min * (team_dreb + opp_oreb)
        )
    else:
        adv_stats['dreb_pct'] = None
    
    # Total Rebound Percentage
    if player_min > 0 and team_min > 0 and (team_reb + opp_reb) > 0:
        adv_stats['reb_pct'] = 100 * (player_reb * (team_min / 5)) / (
            player_min * (team_reb + opp_reb)
        )
    else:
        adv_stats['reb_pct'] = None
    
    # Assist Percentage
    if player_min > 0 and team_min > 0 and (team_fgm - player_fgm) > 0:
        adv_stats['ast_pct'] = 100 * player_ast / (
            ((player_min / (team_min / 5)) * team_fgm) - player_fgm
        )
    else:
        adv_stats['ast_pct'] = None
    
    # ==================== PACE & RATINGS ====================
    
    # Pace (possessions per 48 minutes)
    if team_min > 0:
        adv_stats['pace'] = 48 * ((team_poss + opp_poss) / (2 * (team_min / 5)))
    else:
        adv_stats['pace'] = None
    
    # Offensive Rating (points per 100 possessions)
    adv_stats['off_rating'] = 100 * (team_pts / team_poss) if team_poss > 0 else None
    
    # Defensive Rating (opponent points per 100 possessions)
    adv_stats['def_rating'] = 100 * (opp_pts / opp_poss) if opp_poss > 0 else None
    
    # Net Rating
    if adv_stats['off_rating'] is not None and adv_stats['def_rating'] is not None:
        adv_stats['net_rating'] = adv_stats['off_rating'] - adv_stats['def_rating']
    else:
        adv_stats['net_rating'] = None
    
    # ==================== PIE (Player Impact Estimate) ====================
    
    # PIE numerator (player's contribution)
    pie_player = (
        player_pts + player_fgm + player_ftm - player_fga - player_fta +
        player_dreb + 0.5 * player_oreb + player_ast + player_stl +
        0.5 * player_blk - player_pf - player_tov
    )
    
    # PIE denominator (game total - sum of both teams)
    pie_game = (
        team_pts + team_fgm + (team_pts - team_fgm - team_fta) - team_fga - team_fta +  # team FTM derived
        team_dreb + 0.5 * team_oreb + team_ast + data['team_stl'] + 0.5 * data['team_blk'] -
        data['team_pf'] - team_tov +
        opp_pts + opp_fgm + (opp_pts - opp_fgm - opp_fta) - opp_fga - opp_fta +  # opp FTM derived
        opp_dreb + 0.5 * opp_oreb + data['opp_ast'] + data['opp_stl'] + 0.5 * data['opp_blk'] -
        data['opp_pf'] - opp_tov
    )
    
    adv_stats['pie'] = pie_player / pie_game if pie_game != 0 else None
    
    return adv_stats
```

```python
def compare_with_nba_stats(our_stats, nba_json):
    """
    Compare our calculated stats with NBA's official stats from JSON.
    
    Args:
        our_stats: Dictionary of our calculated advanced stats
        nba_json: List/dict from NBA API (or single dict)
    """
    import json
    
    # Handle if it's a list, take first element
    if isinstance(nba_json, list):
        nba_data = nba_json[0]
    else:
        nba_data = nba_json
    
    # Mapping of NBA stat names to our stat names
    stat_mapping = {
        'EFG_PCT': 'efg_pct',
        'TS_PCT': 'ts_pct',
        'USG_PCT': 'usg_pct',
        'AST_PCT': 'ast_pct',
        'OREB_PCT': 'oreb_pct',
        'DREB_PCT': 'dreb_pct',
        'REB_PCT': 'reb_pct',
        'AST_TOV': 'ast_tov',
        'OFF_RATING': 'off_rating',
        'DEF_RATING': 'def_rating',
        'NET_RATING': 'net_rating',
        'PACE': 'pace',
        'PIE': 'pie'
    }
    
    print(f"\n{'='*80}")
    print(f"STAT COMPARISON - Game: {nba_data.get('GAME_ID')}, Player: {nba_data.get('PLAYER_NAME')}")
    print(f"{'='*80}\n")
    
    print(f"{'STAT':<20} {'NBA VALUE':<20} {'OUR VALUE':<20} {'DIFFERENCE':<20}")
    print(f"{'-'*80}")
    
    comparisons = []
    
    for nba_key, our_key in stat_mapping.items():
        if nba_key in nba_data and our_key in our_stats:
            nba_val = nba_data[nba_key]
            our_val = our_stats[our_key]
            
            # Skip if either is None
            if nba_val is None or our_val is None:
                continue
            
            # Convert NBA percentages from decimal (0-1) to percentage (0-100) if needed
            # NBA stores some as decimals, some as percentages
            if nba_key in ['AST_PCT', 'OREB_PCT', 'DREB_PCT', 'REB_PCT', 'PIE']:
                # These are stored as decimals in NBA data (0.0 - 1.0)
                nba_val_display = nba_val * 100 if nba_val < 2 else nba_val
            elif nba_key in ['EFG_PCT', 'TS_PCT']:
                nba_val_display = nba_val * 100 if nba_val < 2 else nba_val
                our_val_display = our_val * 100 if (our_val and our_val < 2) else our_val
            elif nba_key == 'USG_PCT':
                # USG_PCT appears to be in decimal form too
                nba_val_display = nba_val * 100 if nba_val < 2 else nba_val
            else:
                nba_val_display = nba_val
            
            # Calculate difference
            diff = our_val_display - nba_val_display
            diff_pct = (diff / nba_val_display * 100) if nba_val_display != 0 else 0
            
            # Format based on stat type
            if nba_key in ['EFG_PCT', 'TS_PCT']:
                nba_str = f"{nba_val_display:.3f}%"
                our_str = f"{our_val_display:.3f}%" if our_val_display else "N/A"
                diff_str = f"{diff:+.3f}% ({diff_pct:+.1f}%)"
            elif nba_key in ['AST_PCT', 'OREB_PCT', 'DREB_PCT', 'REB_PCT', 'USG_PCT']:
                nba_str = f"{nba_val_display:.1f}%"
                our_str = f"{our_val:.1f}%" if our_val else "N/A"
                diff_str = f"{diff:+.1f}% ({diff_pct:+.1f}%)"
            elif nba_key == 'PIE':
                nba_str = f"{nba_val_display:.3f}"
                our_str = f"{our_val:.3f}" if our_val else "N/A"
                diff_str = f"{diff:+.3f} ({diff_pct:+.1f}%)"
            elif nba_key == 'AST_TOV':
                nba_str = f"{nba_val:.2f}"
                our_str = f"{our_val:.2f}" if our_val else "N/A"
                diff_str = f"{diff:+.2f} ({diff_pct:+.1f}%)"
            else:  # Ratings, PACE
                nba_str = f"{nba_val_display:.1f}"
                our_str = f"{our_val:.1f}" if our_val else "N/A"
                diff_str = f"{diff:+.1f} ({diff_pct:+.1f}%)"
            
            print(f"{nba_key:<20} {nba_str:<20} {our_str:<20} {diff_str:<20}")
            
            comparisons.append({
                'stat': nba_key,
                'nba': nba_val_display,
                'ours': our_val,
                'diff': diff,
                'diff_pct': diff_pct
            })
    
    # Find stats NBA has that we don't calculate
    print(f"\n{'-'*80}")
    print(f"\nSTATS IN NBA DATA THAT WE DON'T CALCULATE:")
    print(f"{'-'*80}")
    
    nba_only_stats = []
    for key in nba_data.keys():
        if key not in stat_mapping and key not in [
            'index', 'GAME_ID', 'TEAM_ID', 'TEAM_ABBREVIATION', 'TEAM_CITY',
            'PLAYER_ID', 'PLAYER_NAME', 'NICKNAME', 'START_POSITION', 'COMMENT', 'MIN'
        ]:
            nba_only_stats.append(key)
            print(f"  - {key:<25} = {nba_data[key]}")
    
    # Summary
    print(f"\n{'='*80}")
    print(f"SUMMARY:")
    print(f"  Stats compared:              {len(comparisons)}")
    print(f"  Stats NBA has we don't:      {len(nba_only_stats)}")
    
    # Check for large discrepancies
    large_diffs = [c for c in comparisons if abs(c['diff_pct']) > 5]
    if large_diffs:
        print(f"  ⚠️  Large discrepancies (>5%): {len(large_diffs)}")
        for c in large_diffs:
            print(f"      - {c['stat']}: {c['diff_pct']:+.1f}% off")
    else:
        print(f"  ✓ All stats within 5% tolerance")
    
    print(f"{'='*80}\n")
    
    return comparisons, nba_only_stats
```
